```python
- 첫번째 체인 : 나라이름 -> 그 나라에서 가장 유명한 음식
- 두번째 체인 : 음식 -> 음식의 레시피
- 최종 체인 : 나라이름 -> 그 나라에 가장 유명한 음식의 레시피
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model='llama3.2:1b')
output_parser = StrOutputParser()
```

```python
food_prompt = PromptTemplate(
    template="What is the single most famous food in {country}? Return only the name of the food.",
    input_variables=["country"]
)

food_chain = food_prompt | llm | output_parser
```

```python
recipe_prompt = PromptTemplate(
    template="Provide a simple recipe for {food}.",
    input_variables=["food"]
)

recipe_chain = recipe_prompt | llm | output_parser
```

```python
final_chain = {'food': food_chain} | recipe_chain

result = final_chain.invoke({'country': 'Italy'})
print(result)
```
