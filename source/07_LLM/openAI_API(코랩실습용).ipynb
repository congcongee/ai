```python
# 보안 비밀키를 추가 -> 액세스 가능하도록
from openai import OpenAI
from google.colab import userdata
userdata.get('OPEN_KEY')[:5]
client = OpenAI(api_key=userdata.get('OPEN_KEY'))
```

```python
response = client.audio.speech.create(
    model = 'tts-1',
    voice='nova',
    input = 'Hello, OpenAI!'
)
audio_data = response.content
```

```python
# 드라이브 마운트 후, 저장될 폴더로 가서 삼점버튼 클릭하여 경로복사
print(f'생성된 오디오 데이터 크기 : {len(audio_data):,}바이트')
with open('/content/drive/MyDrive/Colab Notebooks/ch5_nova.mp3', 'wb') as audio_file:
    audio_file.write(audio_data)
```
